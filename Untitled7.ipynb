```python
!pip install pyspark -q

import matplotlib.pyplot as plt

from pyspark.sql import SparkSession
from pyspark.sql import functions as F

from pyspark.ml.feature import VectorAssembler, Imputer, StandardScaler
from pyspark.ml.clustering import KMeans
from pyspark.ml.evaluation import ClusteringEvaluator

spark = (
    SparkSession.builder
    .appName("CustomerSegmentation")
    .getOrCreate()
)

print("Spark Version:", spark.version)
```

```python
DATA_PATH = "/content/Customer.csv"

df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(DATA_PATH)
)

print("Total Rows:", df.count())
print("Total Columns:", len(df.columns))

df.show(5)
```

```python
print("Dataset Schema:")
df.printSchema()

print("\nStatistical Summary:")
df.describe().show()

print("\nMissing Values:")

missing = df.select([
    F.sum(
        F.when(F.col(c).isNull(), 1).otherwise(0)
    ).alias(c)
    for c in df.columns
])

missing.show()
```

```python
# Select useful numerical features
feature_columns = [
    "BALANCE",
    "PURCHASES",
    "CASH_ADVANCE",
    "CREDIT_LIMIT",
    "PAYMENTS",
    "TENURE"
]

# Fill missing values using median
imputer = Imputer(
    inputCols=feature_columns,
    outputCols=[col + "_clean" for col in feature_columns],
    strategy="median"
)

df_clean = imputer.fit(df).transform(df)

clean_features = [col + "_clean" for col in feature_columns]

assembler = VectorAssembler(
    inputCols=clean_features,
    outputCol="features"
)

data = assembler.transform(df_clean)

data.select("CUST_ID", "features").show(5, truncate=False)
```

```python
scaler = StandardScaler(
    inputCol="features",
    outputCol="scaled_features",
    withStd=True,
    withMean=True
)

scaler_model = scaler.fit(data)

scaled_data = scaler_model.transform(data)

scaled_data.select(
    "CUST_ID",
    "scaled_features"
).show(5, truncate=False)

print("Feature scaling completed successfully!")
```

```python
kmeans = KMeans(
    k=3,
    seed=42,
    featuresCol="scaled_features",
    predictionCol="cluster"
)

model = kmeans.fit(scaled_data)

predictions = model.transform(scaled_data)

print("Customer Cluster Distribution:")

predictions.groupBy("cluster") \
    .count() \
    .orderBy("cluster") \
    .show()

print("Cluster Centers:")
for i, center in enumerate(model.clusterCenters()):
    print(f"Cluster {i}:", center)
```

```python
evaluator = ClusteringEvaluator(
    featuresCol="scaled_features",
    predictionCol="cluster",
    metricName="silhouette"
)

silhouette = evaluator.evaluate(predictions)

print("Silhouette Score:", round(silhouette, 4))


# Visualise cluster sizes
plot_data = (
    predictions
    .groupBy("cluster")
    .count()
    .orderBy("cluster")
    .toPandas()
)

plt.figure(figsize=(7, 5))

plt.bar(
    plot_data["cluster"].astype(str),
    plot_data["count"]
)

plt.title("Customer Cluster Distribution")
plt.xlabel("Customer Cluster")
plt.ylabel("Number of Customers")

plt.show()


print("""
====================================================
WEEK 5 — CUSTOMER SEGMENTATION USING PYSPARK
====================================================

PROJECT COMPLETED SUCCESSFULLY

DATA PROCESSING:
• Spark DataFrames
• Missing value handling
• Median imputation
• Feature vector creation
• Feature scaling

MACHINE LEARNING:
• K-Means Clustering
• 3 Customer Segments

EVALUATION:
• Silhouette Score
• Cluster distribution analysis

CONCLUSION:
Customer financial behaviour was analysed using
PySpark and grouped into three distinct clusters.
K-Means clustering identified patterns based on
balance, purchases, cash advances, credit limits,
payments and customer tenure.

====================================================
WEEK 5 COMPLETE
====================================================
""")
```
